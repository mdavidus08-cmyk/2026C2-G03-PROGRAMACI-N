# Problema #1: Calculadora de propina inteligente

## Problema

Solicitando el valor total de la cuenta y el porcentaje de propina que deseo dejar. Calcula el monto de la propina y el total final para pagar, mostrando los valores con formato calro 

## Entrada

- Valor total de la cuenta (numero decimal positivo)
- porcentaje de propina (numero decimal positivo)
## Proceso

- Validar que los valores ingresados sean 
numeros mayores que cero 

-Formula

propina = total cuenta *
(porcentaje / 100)
total_final = total_cuenta + propina 

## Salida esperada

- Muestra el monto de la cuenta, la propina calculada y total final, con dos decimales. Si hay error, muestra un mensaje claro.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|-----------|------------------|--------------------|--------------------|
| Caso normal | Cuenta=100, Propina=15  | Propina=$15,00,Total=115.00 | Propina=$15,00, Total a pagar= $115,00 |
| Caso límite | Cuenta=20, Propina=0 | Propina= $0,00, Total a pagar= $20,00 | Propina= $0,00, Total a pagar= $20,00 |
| Caso inválido | Cuenta= -50 o "cien", propina= 0  | Mensaje: Error: Los valores deben ser mayores a cero o Error: Ingrese solo numeros validos | Mensaje: Error: Los valores deben ser mayores a cero o Error: Ingrese solo numeros validos |


```python
# Calculadora de propina inteligente
try:
    # Entrada
    total_cuenta = float(input("Ingrese el valor total de la cuenta: $"))
    porcentaje_propina = float(input("Ingrese el porcentaje de propina (ej: 10 para 10%): "))

    # Validación
    if total_cuenta <= 0 or porcentaje_propina < 0:
        print(" Error: Los valores deben ser mayores a cero.")
    else:
        # Cálculo
        propina = total_cuenta * (porcentaje_propina / 100)
        total_final = total_cuenta + propina

        # Salida
        print("\n--- Resumen de la cuenta ---")
        print(f"Total de la cuenta: ${total_cuenta:.2f}")
        print(f"Propina ({porcentaje_propina:.1f}%): ${propina:.2f}")
        print(f"Total a pagar: ${total_final:.2f}")

except ValueError:
    print(" Error: Ingrese solo números válidos.")
```

# Problema #2: Detector de número par/impar

## Problema

Recibe unnumero entero ingresado por el usuario y determina si es par o impar, mostrando un mensaje claro con el resultado. Ademas valida que el dato ingresado sea un numero valido

## Entrada

- Un numero entero (positivo, negativo, o cero)

## Proceso

- Se verfica que el valor ingresado sea un numero entero valido 

- Se usa el operador modulo %: si el resto de dividir el numero entre 2 es igual a 0, el numero es par; em caso contrario es impar

- Se maneja el error si el usuario ingresa texto o valores no enteros 

## Salida esperada

- Muestra el mensaje: El numero X es PAR o El numero X es IMPAR 

- Si el dato es incorrecto, muestra: Error: Ingrese solamente numeros enteros validos 

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|-----------|------------------|--------------------|--------------------|
| Caso normal | 7 | El numero 7 es IMPAR | El numero 7 es IMPAR |
| Caso límite | 0 | El mumero 0 es PAR| El mumero 0 es PAR |
| Caso inválido | "Hola" o 5.8 | Error ingrese solo numeros enteros validos | Error ingrese solo numeros enteros validos |


```python
# Detector de número par o impar
try:
    # Entrada
    numero = int(input("Ingrese un número entero: "))

    # Proceso y clasificación
    if numero % 2 == 0:
        print(f" El número {numero} es PAR")
    else:
        print(f" El número {numero} es IMPAR")

except ValueError:
    print(" Error: Ingrese solo números enteros válidos.")
```

# Problema #3: Clasificador de edad

## Problema

Solicita la edad de una persona y la clasifica en tres categorías: menor de edad, adulto o adulto mayor. Valida que la edad ingresada sea un valor numérico válido y no negativo.

## Entrada

- Un número entero que representa la edad de la persona.

## Proceso

- Se verifica que el valor ingresado sea un número entero y mayor o igual a 0.

- Se aplica la clasificación:

- Si es menor a 18 → Menor de edad

- Si está entre 18 y 59 años → Adulto

- Si es 60 años o más → Adulto mayor

- Se muestra un mensaje de error si se ingresa texto o valores negativos.

## Salida esperada

- Muestra la categoría correspondiente: Con X años es Menor de edad / Adulto / Adulto mayor.

- En caso de error: Error: Ingrese una edad válida, mayor o igual a 0.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|-----------|------------------|--------------------|--------------------|
| Caso normal | 25 | Con 25 años es ADULTO | Con 25 años es ADULTO  |
| Caso límite | 18 o 60 | Con 18 años es ADULTO, Con 60 años es ADULTO MAYOR | Con 18 años es ADULTO, Con 60 años es ADULTO MAYOR |
| Caso inválido | -7 o "veinte" | Error: ingrese una edad valida, mayor o igual a 0 | Error: ingrese una edad valida, mayor o igual a 0  |


```python
# Clasificador de edad
try:
    # Entrada
    edad = int(input("Ingrese su edad: "))

    # Validación y clasificación
    if edad < 0:
        print(" Error: La edad no puede ser negativa.")
    elif edad < 18:
        print(f"Con {edad} años: Es MENOR DE EDAD")
    elif edad < 60:
        print(f"Con {edad} años: Es ADULTO")
    else:
        print(f"Con {edad} años: Es ADULTO MAYOR")

except ValueError:
    print(" Error: Ingrese solo números enteros válidos.")
```

# Problema #4: Validador de compra con descuento

## Problema

Solicita el monto total de una compra y si el cliente es frecuente. Aplica descuentos según las reglas:

- Más de $100 y cliente frecuente → 15% de descuento

- Más de $100 pero no es frecuente → 10% de descuento

- * Menos o igual a $100 → sin descuento
Muestra el monto original, el descuento aplicado y el total final a pagar, validando que los datos ingresados sean correctos

## Entrada

- Monto de la compra: número positivo (puede tener decimales)

- Condición de cliente frecuente: texto si o no

## Proceso

- Se limpia el texto ingresado y se convierte a minúsculas para evitar errores de escritura

- Se valida que el monto sea mayor a 0 y la respuesta sea solo si o no.

- Se calcula el descuento y el total final según las reglas establecidas.

- Se usa try-except para capturar errores si se ingresa texto en lugar de números.

## Salida esperada

- Muestra resumen: monto original, porcentaje de descuento, valor ahorrado y total final.

- Si hay error: mensaje claro indicando qué dato es incorrecto.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|-----------|------------------|--------------------|--------------------|
| Caso normal | Monto = 200, Frecuente = si | Monto=$200,00, Descuento=15%($30,00), Total a pagar=$170.00 | Monto=$200,00, Descuento=15%($30,00), Total a pagar=$170.00 |
| Caso límite | Monto=100, Frecuente = si  | Monto=$100,00, Descuento=0%($0,00), Total a pagar=$100.00 |  Monto=$100,00, Descuento=0%($0,00), Total a pagar=$100.00 |
| Caso inválido | Monto = -50 / "cien", Frecuente = Talvez | Error: El monto debe de ser mayor a cero /Responda solo con "si" o "no" | Error: El monto debe de ser mayor a cero /Responda solo con "si" o "no" |


```python
# Validador de compra con descuento
try:
    # Entrada y limpieza de datos
    monto = float(input("Ingrese el monto total de la compra: $ "))
    frecuente = input("¿Es cliente frecuente? (si/no): ").strip().lower()

    # Validación
    if monto <= 0:
        print("Error: El monto debe ser mayor a cero.")
    elif frecuente not in ("si", "no"):
        print("Error: Responda solo con 'si' o 'no'.")
    else:
        # Cálculo del descuento
        if monto > 100 and frecuente == "si":
            descuento = 0.15
            porcentaje = "15%"
        elif monto > 100 and frecuente == "no":
            descuento = 0.10
            porcentaje = "10%"
        else:
            descuento = 0
            porcentaje = "0%"

        ahorro = monto * descuento
        total_final = monto - ahorro

        
        print("\n--- Resumen de la compra ---")
        print(f"Monto original: $ {monto:.2f}")
        print(f"Descuento aplicado: {porcentaje} → $ {ahorro:.2f}")
        print(f"Total a pagar: $ {total_final:.2f}")

except ValueError:
    print("Error: El monto debe ser un número válido.")
```
