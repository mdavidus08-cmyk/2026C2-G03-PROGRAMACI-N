# Caso 1: Control de ventas diarias de una soda

## Problema
Una soda ubicada en San Pedro registra sus ventas de lunes a viernes. El programa debe calcular el total semanal, el promedio diario y clasificar el resultado según una meta.

## Entrada / datos
Use esta lista base:

```python
ventas_soda = [85000, 92000, 78000, 110000, 97000]
meta_semanal = 450000
```

## Proceso esperado

- Calcular total semanal.
- Calcular promedio diario.
- Comparar el total con la meta.
- Clasificar:
  - `Meta alcanzada` si el total es mayor o igual a la meta.
  - `Cerca de la meta` si el total alcanza al menos el 80% de la meta.
  - `Requiere atención` si no alcanza el 80%.

## Salida esperada
Debe mostrar total, promedio, porcentaje de cumplimiento y estado final.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|---|---|---|---|
| Caso normal | `[85000, 92000, 78000, 110000, 97000]`, meta `450000` | Meta alcanzada | Meta alcanzada |
| Caso límite |[90000, 90000, 90000, 90000, 90000], meta 450000 | Meta alcanzada | Meta alcanzada |
| Caso especial | [50000, 60000, 55000, 70000, 65000], meta 450000 | Requiere atención |Requiere atención |

````python
# Caso 1 - Escriba su solución aquí

ventas_soda = [85000, 92000, 78000, 110000, 97000]
meta_semanal = 450000


def calcular_porcentaje_cumplimiento(total_semanal, meta_semanal):
    return total_semanal / meta_semanal * 100


total_semanal = 0

for venta in ventas_soda:
    total_semanal = total_semanal + venta

promedio_diario = total_semanal / len(ventas_soda)
porcentaje_cumplimiento = calcular_porcentaje_cumplimiento(total_semanal, meta_semanal)

if total_semanal >= meta_semanal:
    estado_final = "Meta alcanzada"
elif total_semanal >= meta_semanal * 0.80:
    estado_final = "Cerca de la meta"
else:
    estado_final = "Requiere atención"

print("REPORTE DE VENTAS SEMANALES")
print("Total semanal:", total_semanal)
print("Promedio diario:", promedio_diario)
print("Porcentaje de cumplimiento:", format(porcentaje_cumplimiento, ".2f") + "%")
print("Estado final:", estado_final)

````

# Caso 2: Asistencia a una actividad cultural

## Problema
Una actividad cultural en Costa Rica registra asistencia por bloques horarios. El programa debe calcular la asistencia promedio y clasificar el nivel de ocupación.

## Entrada / datos
Use esta lista base:

```python
asistencias = [120, 180, 210, 250, 230]
capacidad = 300
```

## Proceso esperado

- Recorrer la lista con un ciclo `for`.
- Calcular asistencia total y promedio.
- Contar cuántos bloques superaron el 70% de capacidad.
- Clasificar el promedio de ocupación:
  - `Alta ocupación` si el promedio es mayor o igual al 80% de la capacidad.
  - `Ocupación media` si el promedio es mayor o igual al 50%.
  - `Baja ocupación` si el promedio es menor al 50%.

## Salida esperada
Debe mostrar total de asistentes, promedio, bloques sobre 70% y estado de ocupación.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|---|---|---|---|
| Caso normal | `[120, 180, 210, 250, 230]`, capacidad `300` | Ocupación media o alta según cálculo |Ocupación media |
| Caso límite | [240, 240, 240, 240, 240], capacidad 300 | Alta ocupación |Alta ocupación |
| Caso especial | [40, 80, 90, 100, 110], capacidad 300 | Baja ocupación |Baja ocupación |

````python
# Caso 2 - Escriba su solución aquí

asistencias = [120, 180, 210, 250, 230]
capacidad = 300

total_asistentes = 0
bloques_sobre_70 = 0

for asistencia in asistencias:
    total_asistentes = total_asistentes + asistencia

    if asistencia > capacidad * 0.70:
        bloques_sobre_70 = bloques_sobre_70 + 1

promedio_asistencia = total_asistentes / len(asistencias)
porcentaje_promedio = promedio_asistencia / capacidad * 100

if promedio_asistencia >= capacidad * 0.80:
    estado_ocupacion = "Alta ocupación"
elif promedio_asistencia >= capacidad * 0.50:
    estado_ocupacion = "Ocupación media"
else:
    estado_ocupacion = "Baja ocupación"

print("REPORTE DE ASISTENCIA")
print("Total de asistentes:", total_asistentes)
print("Promedio de asistencia:", promedio_asistencia)
print("Porcentaje promedio de ocupación:", format(porcentaje_promedio, ".2f") + "%")
print("Bloques sobre 70% de capacidad:", bloques_sobre_70)
print("Estado de ocupación:", estado_ocupacion)
# Complete el análisis
````

# Caso 3: Precios de productos de feria

## Problema
Una persona compara precios de un mismo producto en diferentes ferias del agricultor. El programa debe analizar los precios y detectar si hay mucha variación.

## Entrada / datos
Use esta lista base:

```python
producto = "Tomate"
precios = [900, 850, 1000, 950, 1200]
```

## Proceso esperado

- Calcular precio promedio.
- Identificar precio mínimo y máximo.
- Calcular rango de precios: `máximo - mínimo`.
- Clasificar la variación:
  - `Variación alta` si el rango es mayor o igual a 300.
  - `Variación moderada` si el rango es mayor o igual a 150.
  - `Variación baja` si el rango es menor a 150.
- Usar un ciclo para contar cuántos precios están por encima del promedio.

## Salida esperada
Debe mostrar producto, promedio, mínimo, máximo, rango, precios sobre promedio y clasificación.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|---|---|---|---|
| Caso normal | `[900, 850, 1000, 950, 1200]` | Variación alta | Variación alta |
| Caso límite | [900, 1000, 1100, 1200] | Variación alta | Variación alta |
| Caso especial | [950, 950, 950, 950] | Variación baja | Variación baja |

````python
# Caso 3 - Escriba su solución aquí
producto = "Tomate"
precios = [900, 850, 1000, 950, 1200]

suma_precios = 0
precio_minimo = precios[0]
precio_maximo = precios[0]
precios_sobre_promedio = 0

for precio in precios:
    suma_precios = suma_precios + precio

    if precio < precio_minimo:
        precio_minimo = precio

    if precio > precio_maximo:
        precio_maximo = precio

precio_promedio = suma_precios / len(precios)
rango_precios = precio_maximo - precio_minimo

for precio in precios:
    if precio > precio_promedio:
        precios_sobre_promedio = precios_sobre_promedio + 1

if rango_precios >= 300:
    clasificacion = "Variación alta"
elif rango_precios >= 150:
    clasificacion = "Variación moderada"
else:
    clasificacion = "Variación baja"

print("REPORTE DE PRECIOS")
print("Producto:", producto)
print("Precio promedio:", precio_promedio)
print("Precio mínimo:", precio_minimo)
print("Precio máximo:", precio_maximo)
print("Rango de precios:", rango_precios)
print("Precios sobre promedio:", precios_sobre_promedio)
print("Clasificación:", clasificacion)
````

Use esta lista de diccionarios:

sedes = [

   - nombre;
   - provincia;

   - porcentaje de cumplimiento;
   - estado.
6. Crear un `set` con las provincias presentes.
7. Crear una lista de tuplas `(nombre, total)` para guardar el ranking base.
8. Identificar la sede con mayor venta total.

## Clasificación requerida

- `Meta alcanzada`: total mayor o igual a la meta.
- `Cerca de la meta`: total mayor o igual al 80% de la meta.
- `Requiere atención`: total menor al 80% de la meta.

## Salida esperada
El reporte debe ser claro y comprensible para una persona que no está viendo el código.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|---|---|---|---|
| Caso normal | datos base | Reporte para las 3 sedes | Meta alcanzada |
| Caso límite | Una sede con ventas [90000, 90000, 90000, 90000, 90000] y meta 450000 | Meta alcanzada | Meta alcanzada |
| Caso especial | Una sede con ventas [10000, 12000, 9000, 11000, 10000] y meta 300000 | Requiere atención | Requiere atención |

````python
# Caso 4 - Reto integrador

def calcular_total(ventas):
    total = 0

    for venta in ventas:
        total = total + venta

    return total


def calcular_promedio(ventas):
    total = calcular_total(ventas)
    promedio = total / len(ventas)

    return promedio


def clasificar_sede(total, meta):
    if total >= meta:
        estado = "Meta alcanzada"
    elif total >= meta * 0.80:
        estado = "Cerca de la meta"
    else:
        estado = "Requiere atención"

    return estado


sedes = [
    {
        "nombre": "Soda San Pedro",
        "provincia": "San José",
        "tipo": "Alimentación",
        "ventas": [85000, 92000, 78000, 110000, 97000],
        "meta": 450000,
    },
    {
        "nombre": "Panadería Cartago",
        "provincia": "Cartago",
        "tipo": "Panadería",
        "ventas": [65000, 70000, 72000, 69000, 75000],
        "meta": 380000,
    },
    {
        "nombre": "Cafetería Liberia",
        "provincia": "Guanacaste",
        "tipo": "Cafetería",
        "ventas": [98000, 105000, 99000, 120000, 115000],
        "meta": 500000,
    },
]

provincias_presentes = set()
ranking_base = []

mayor_venta_total = 0
sede_mayor_venta = ""

print("REPORTE SEMANAL RED EMPRENDECR")
print()

for sede in sedes:
    nombre = sede["nombre"]
    provincia = sede["provincia"]
    tipo = sede["tipo"]
    ventas = sede["ventas"]
    meta = sede["meta"]

    total_semanal = calcular_total(ventas)
    promedio_diario = calcular_promedio(ventas)
    porcentaje_cumplimiento = total_semanal / meta * 100
    estado = clasificar_sede(total_semanal, meta)

    provincias_presentes.add(provincia)
    ranking_base.append((nombre, total_semanal))

    if total_semanal > mayor_venta_total:
        mayor_venta_total = total_semanal
        sede_mayor_venta = nombre

    print("Sede:", nombre)
    print("Provincia:", provincia)
    print("Tipo de negocio:", tipo)
    print("Total semanal:", total_semanal)
    print("Promedio diario:", promedio_diario)
    print("Porcentaje de cumplimiento:", format(porcentaje_cumplimiento, ".2f") + "%")
    print("Estado:", estado)
    print()

print("RESUMEN GENERAL")
print("Provincias presentes:", provincias_presentes)
print("Ranking base:", ranking_base)
print("Sede con mayor venta total:", sede_mayor_venta)
print("Mayor venta total:", mayor_venta_total)

# 1. Defina las funciones


# 2. Recorra las sedes y genere el reporte


# 3. Muestre provincias presentes, ranking base y mejor sede
````
